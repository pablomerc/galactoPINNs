# Catalog $\sigma(a_{\mathrm{LOS}})$ distribution

Per-pulsar measurement noise from the 51 Donlon+2025 pulsars (PB-else-PS selection). Histogram + Gaussian KDE; vertical lines mark the constant noise levels used in the experiment (~10th / 50th / 75th percentiles).

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde

# Embedded literals from noisy_datasets / GUIDE_1 (data/ is gitignored).
CATALOG_SIGMAS_MMSYR = np.asarray((
    0.118371, 0.326056, 0.33556, 0.339927, 0.420831, 0.523387,
    0.532496, 0.554591, 0.576837, 0.670016, 0.726771, 0.745114,
    0.763416, 0.770922, 0.851857, 0.915736, 0.920055, 1.01904,
    1.0872, 1.36575, 1.46129, 1.46334, 1.50857, 1.64275,
    1.68204, 1.69031, 1.82323, 1.87895, 1.92658, 2.06964,
    2.41148, 2.77526, 2.79087, 2.81604, 3.15476, 3.16167,
    3.64418, 4.12575, 4.4002, 4.53605, 5.95132, 6.07849,
    6.35578, 6.93263, 7.12839, 7.36009, 8.57286, 13.0403,
    13.5344, 18.2143, 18.2143,
))

# Constant noise levels ≈ 10th / 50th / 75th percentiles of the catalog.
LEVELS = {
    "s0.5 (≈10th)": 0.52,
    "s1.7 (≈50th)": 1.69,
    "s4.3 (≈75th)": 4.26,
}

sig = CATALOG_SIGMAS_MMSYR
print(f"N = {len(sig)}")
print(f"min / max = {sig.min():.3f} / {sig.max():.3f} mm/s/yr")
for q in (10, 25, 50, 75, 90):
    print(f"p{q:02d} = {np.percentile(sig, q):.3f}")
```

```python
kde = gaussian_kde(sig)
x = np.linspace(0.0, sig.max() * 1.05, 400)
density = kde(x)

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.hist(sig, bins=20, density=True, color="#5b7c99", edgecolor="white",
        linewidth=0.8, alpha=0.75, label="catalog histogram")
ax.plot(x, density, color="#1a1a1a", lw=2.0, label="Gaussian KDE")

colors = ("#2a9d8f", "#e9c46a", "#e76f51")
for (label, val), c in zip(LEVELS.items(), colors):
    ax.axvline(val, color=c, ls="--", lw=1.4, label=f"{label} = {val}")

ax.set_xlabel(r"$\sigma(a_{\mathrm{LOS}})$ [mm/s/yr]")
ax.set_ylabel("density")
ax.set_title("Donlon+2025 pulsar timing noise (51 pulsars)")
ax.legend(frameon=False, fontsize=9)
ax.set_xlim(0, None)
ax.spines[["top", "right"]].set_visible(False)
fig.tight_layout()
plt.show()
```
